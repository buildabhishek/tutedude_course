```python
# Section 1: Imports

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style("whitegrid")
```

```python
# Section 2: Load Dataset

df = pd.read_csv("Titanic-Dataset.csv")
```

```python
# Section 3: Data Exploration


df.head()
df.info()
df.describe()
df.shape
df.columns
df.isnull().sum()
```

```python
# Section 4: Cleaning

df["Age"] = df["Age"].fillna(df["Age"].median())

df["Embarked"] = df["Embarked"].fillna(
    df["Embarked"].mode()[0]
)
```

```python
# TASK 1- Relational Plot
sns.relplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Sex"
)

plt.show()
```

```python
# Scatter Version

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Sex"
)

plt.tight_layout()
plt.show()
```

```python
# TASK 2 - Line Plot

plt.figure(figsize=(10,6))

sns.lineplot(
    data=df,
    x="Pclass",
    y="Fare"
)

plt.show()
```

```python
# Faceted Line Plot

sns.relplot(
    data=df,
    x="Pclass",
    y="Fare",
    kind="line",
    col="Sex"
)

plt.show()
```

```python
# TASK 3 - Distribution Plots , Histogram

plt.figure(figsize=(10,6))

sns.histplot(
    df["Age"],
    kde=False
)

plt.show()
```

```python
# KDE

plt.figure(figsize=(10,6))

sns.kdeplot(df["Age"])

plt.show()
```

```python
# Rug Plot

plt.figure(figsize=(10,6))

sns.rugplot(df["Age"])

plt.show()
```

```python
# Histogram + KDE + Rug

plt.figure(figsize=(10,6))

sns.histplot(
    df["Age"],
    kde=True
)

sns.rugplot(df["Age"])

plt.show()
```

```python
# TASK 4 - Bivariate Distribution
# KDE

plt.figure(figsize=(10,6))

sns.kdeplot(
    data=df,
    x="Age",
    y="Fare",
    fill=True
)

plt.show()
```

```python
# Histogram

plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="Age",
    y="Fare"
)

plt.show()
```

```python
# TASK 5 - Matrix Plots
# Pairplot

sns.pairplot(
    df[
        ["Age",
         "Fare",
         "Pclass",
         "SibSp"]
    ]
)

plt.show()
```

```python
# Heatmap

plt.figure(figsize=(10,6))

corr = df[
    ["Age","Fare","Pclass","SibSp"]
].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.show()
```

```python
# TASK 6 - Categorical Plots
# Barplot

plt.figure(figsize=(10,6))

sns.barplot(
    data=df,
    x="Sex",
    y="Fare"
)

plt.show()
```

```python
# Boxplot

plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="Sex",
    y="Fare"
)

plt.show()
```

```python
# Violinplot

plt.figure(figsize=(10,6))

sns.violinplot(
    data=df,
    x="Sex",
    y="Fare"
)

plt.show()
```

```python
# Countplot

plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    x="Sex"
)

plt.show()
```

```python
# TASK 7 - Regression Plots
# Regplot

plt.figure(figsize=(10,6))

sns.regplot(
    data=df,
    x="Age",
    y="Fare"
)

plt.show()
```

```python
# lmplot

sns.lmplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Sex"
)

plt.show()
```

```python
# TASK 8 - Multi-Plots
# FacetGrid

g = sns.FacetGrid(
    df,
    col="Sex"
)

g.map(
    sns.scatterplot,
    "Age",
    "Fare"
)

plt.show()
```

```python
# Catplot

sns.catplot(
    data=df,
    x="Pclass",
    y="Fare",
    kind="box"
)

plt.show()
```

```python
# Displot

sns.displot(
    data=df,
    x="Age",
    kde=True
)

plt.show()
```

# Final Findings

1. Most passengers were between 20 and 40 years old.
2. Fare distribution is heavily right-skewed.
3. Several fare outliers exist.
4. Passenger class influences fare amount.
5. Female passengers generally paid higher fares.
